```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load csv file into a pandas dataframe
df = pd.read_csv(r'C:\Users\matia\Desktop\report\step2\concatenated\OSLO\start_end_station_df.csv')
print(df.head(10))

# Print the data types of each column
print(df.dtypes)
```

```python
# Count unique Start and Stopp combinations
trip_counts = df.groupby(['Start', 'Stopp']).size().reset_index(name='Count')

# Sort by count in descending order and get the top 5
top_5_trips = trip_counts.sort_values(by='Count', ascending=False).head(5)

# Print the top 5 trips with their counts
print(top_5_trips)

# Visualization
plt.figure(figsize=(10, 6))
plt.barh(
    [f"{row['Start']} -> {row['Stopp']} ({row['Count']})" for _, row in top_5_trips.iterrows()],
    top_5_trips['Count']
)
plt.xlabel('Antall bysykkel-turer')
plt.ylabel('Tur = (Start -> Stopp)')
plt.title('Topp fem Oslo Bysykkel-turer i 2024')
plt.gca().invert_yaxis()  # Invert y-axis to show the most frequent trip on top
plt.show()
```
