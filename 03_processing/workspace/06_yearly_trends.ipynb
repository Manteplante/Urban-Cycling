# 06 Yearly Trends

Build a simple yearly trip-count table per city for the Streamlit Yearly Trends chapter.

Export:
- yearly_trends_city_year.csv

```python
import sys
import pandas as pd
import plotly.express as px

sys.path.insert(0, '..')
from utils import load_app_ready, export_df
```

```python
df = load_app_ready()
df.shape
```

```python
required = ['city_name', 'year', 'trip_id']
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f'Missing required columns for yearly trends: {missing}')

yearly = (
    df.dropna(subset=['city_name', 'year'])
    .assign(year=lambda x: pd.to_numeric(x['year'], errors='coerce'))
    .dropna(subset=['year'])
    .assign(year=lambda x: x['year'].astype(int))
    .groupby(['city_name', 'year'], as_index=False)
    .agg(trips=('trip_id', 'count'))
    .sort_values(['city_name', 'year'])
    .reset_index(drop=True)
)

yearly.head()
```

```python
fig = px.bar(
    yearly,
    x='year',
    y='trips',
    color='city_name',
    barmode='group',
    title='Trips per Year by City',
    labels={'year': 'Year', 'trips': 'Trips', 'city_name': 'City'},
)
fig.update_layout(plot_bgcolor='white', legend_title_text='City')
fig.update_xaxes(type='category')
fig.show()
```

```python
export_df('yearly_trends_city_year', yearly)
print('Exported yearly trends table.')
```
